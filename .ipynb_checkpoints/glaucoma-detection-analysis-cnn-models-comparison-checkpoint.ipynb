## Dataset Structure

```python
def print_folder_structure(root_dir):
    file_name = []
    file_count = []

    for root, dirs, files in os.walk(root_dir): # Calculate the depth of the current directory for indentation
        depth = root[len(root_dir):].count(os.sep)
        indent = " " * 4 * depth
        
        # Count all files in the current directory and its subdirectories
        total_files = sum(len(files) for _, _, files in os.walk(root))
        
        print(f"{indent}[{os.path.basename(root)}] - {total_files} files")

        file_name.append(f'{os.path.basename(root)}')
        file_count.append(total_files)
    return file_name, file_count
        

# Define the path to the root dataset directory
dataset_path = "/kaggle/input/glaucoma-large/glaucoma_dataset"

# Call the function
print("Folder Structure and File Count:\n")
file_names, file_counts = print_folder_structure(dataset_path)
    
```

# Data Preparation

```python
# Paths to datasets
train_dataset_path = '/kaggle/input/glaucoma-large/glaucoma_dataset/train'
validation_dataset_path = '/kaggle/input/glaucoma-large/glaucoma_dataset/test'
test_dataset_path = '/kaggle/input/glaucoma-large/glaucoma_dataset/val'
```

```python
# Define image size and batch size
IMG_SIZE = (224, 224)
BATCH_SIZE = 16
```

```python
# Image transformations for training and validation
train_transform = transforms.Compose([
    transforms.Resize(IMG_SIZE),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(10),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

test_transform = transforms.Compose([
    transforms.Resize(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])
```

```python
# Load datasets
train_dataset = datasets.ImageFolder(root=train_dataset_path, transform=train_transform)
val_dataset = datasets.ImageFolder(root=validation_dataset_path, transform=test_transform)
test_dataset = datasets.ImageFolder(root=test_dataset_path, transform=test_transform)
```

```python
# the size of the new datasets
print(f"Train dataset size: {len(train_dataset)}")
print(f"Validation dataset size: {len(val_dataset)}")
print(f"Test dataset size: {len(test_dataset)}")
```

```python
train_samples =8000
test_samples =1000
val_samples =1000
```

```python
# Load limited images from datasets
train_subset, _ = random_split(train_dataset, [train_samples, len(train_dataset) - train_samples])
val_subset, _ = random_split(val_dataset, [val_samples, len(val_dataset) - val_samples])
test_subset, _ = random_split(test_dataset, [test_samples, len(test_dataset) - test_samples])
```

```python
# Data loaders
train_loader = torch.utils.data.DataLoader(train_subset, batch_size=BATCH_SIZE, shuffle=True)
val_loader = torch.utils.data.DataLoader(val_subset, batch_size=BATCH_SIZE, shuffle=False)
test_loader = torch.utils.data.DataLoader(test_subset, batch_size=BATCH_SIZE, shuffle=False)
```

```python
# Class labels
labels = {v: k for k, v in train_dataset.class_to_idx.items()}
print("Class Labels:", labels)
```

```python
# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
```

```python

def test_with_metrics(model, dataloader):
    y_true = []
    y_pred = []
    model.eval()
    # Initialize the progress bar
    loader = tqdm(dataloader, desc="Testing", leave=False)

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device).float().unsqueeze(1)

            # Forward pass
            outputs = model(images)
            predictions = (outputs >= 0.5).float()

            # Append true labels and predicted labels
            y_true.extend(labels.cpu().numpy())
            y_pred.extend(predictions.cpu().numpy())

    # Calculate the confusion matrix
    cm = confusion_matrix(y_true, y_pred)

    # Calculate additional metrics
    accuracy = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred)
    recall = recall_score(y_true, y_pred)
    f1 = f1_score(y_true, y_pred)

    # Plot the confusion matrix as a heatmap
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu', xticklabels=["Normal", "Glaucoma"], yticklabels=["Normal", "Glaucoma"])
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title("Confusion Matrix")

    # Display the calculated metrics on the plot
    plt.text(0.5, -0.2, f"Accuracy: {accuracy:.4f}\nRecall: {recall:.4f} & Precision: {precision:.4f}",
             fontsize=12, ha='center', va='center', transform=plt.gca().transAxes)

    plt.show()

    # Print out the metrics
    print(f"Test Accuracy:  {accuracy:.4f}")
    print(f"Precision:      {precision:.4f}")
    print(f"Recall:         {recall:.4f}")
    print(f"F1 Score:       {f1:.4f}")

    return {'accuracy':accuracy*100, 'precision':precision*100, 'recall':recall*100, 'F1 Score':f1*100}
```

## VGG19

### Model is trained [here](https://www.kaggle.com/code/developersampad/glaucoma-detection-using-vgg19)

```python
# Load VGG19 model
base_model = models.vgg19(weights=None)

# Modify the classifier for binary classification
base_model.classifier = nn.Sequential(
    nn.Linear(25088, 512),
    nn.ReLU(),
    # nn.Dropout(0.5),
    nn.Linear(512, 1),
    nn.Sigmoid()
)
vgg_path = '/kaggle/input/trained-for-glaucoma-detection/pytorch/default/1/vgg19_glaucoma_model.pth'

base_model.load_state_dict(torch.load(vgg_path, map_location=device, weights_only=True))

# Move the model to the device
vgg19_model = base_model.to(device)

# Training Plots
plt.figure(figsize=(12,5))
t_plot = cv2.imread('/kaggle/input/training-plots/vgg19_plot_training.png')
t_plot = cv2.cvtColor(t_plot, cv2.COLOR_BGR2RGB)
plt.imshow(t_plot)
plt.title('VGG19 Training Plots')
plt.axis('off')
plt.show()
```

```python
vgg19_tm = test_with_metrics(vgg19_model, test_loader)
models_dict = {'VGG19':vgg19_tm}
```
