# Model

```python
# Load Vision Transformer model
base_model = models.vit_b_16(pretrained=True)

# Freeze base model layers
for param in base_model.parameters():
    param.requires_grad = False

# Modify the classifier for binary classification
base_model.heads = nn.Sequential(
    nn.Linear(base_model.hidden_dim, 512),
    nn.ReLU(),
    # nn.Dropout(0.5),
    nn.Linear(512, 1),
    nn.Sigmoid()
)

# Move the model to the device
model = base_model.to(device)
```

```python
# Define loss function and optimizer
criterion = nn.BCELoss()  # Binary cross-entropy loss
optimizer = optim.Adam(model.heads.parameters(), lr = 0.001)

print(model)
```

# Training

### Defining Training Function

```python
def train(model, dataloader, criterion, optimizer):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0

    # Initialize the progress bar
    loader = tqdm(dataloader, desc="Training", leave=False)

    for images, labels in loader:
        images, labels = images.to(device), labels.to(device).float().unsqueeze(1)

        # Forward pass
        outputs = model(images)
        loss = criterion(outputs, labels)

        # Backward pass and optimization
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        # Statistics
        running_loss += loss.item() * images.size(0)
        predicted = (outputs >= 0.5).float()
        correct += (predicted == labels).sum().item()
        total += labels.size(0)

        # Update the loader description with current loss and accuracy
        current_loss = running_loss / total
        current_accuracy = correct / total
        loader.set_postfix(loss=current_loss, accuracy=current_accuracy)

    epoch_loss = running_loss / total
    accuracy = correct / total
    return epoch_loss, accuracy
```

### Defining Evaluation on Validation Function

```python
def validate(model, dataloader, criterion):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0

    # Initialize the progress bar
    loader = tqdm(dataloader, desc="Validation", leave=False)

    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device).float().unsqueeze(1)

            # Forward pass
            outputs = model(images)
            loss = criterion(outputs, labels)

            # Statistics
            running_loss += loss.item() * images.size(0)
            predicted = (outputs >= 0.5).float()
            correct += (predicted == labels).sum().item()
            total += labels.size(0)

            # Update the loader description with current loss and accuracy
            current_loss = running_loss / total
            current_accuracy = correct / total
            loader.set_postfix(loss=current_loss, accuracy=current_accuracy)

    epoch_loss = running_loss / total
    accuracy = correct / total
    return epoch_loss, accuracy
```

### Training Operation

```python
epochs = 40
train_losses, val_losses = [], []
train_accuracies, val_accuracies = [], []

for epoch in range(epochs):
    # Patch: Remap labels in train/validate functions for binary classification
    def binary_train(model, dataloader, criterion, optimizer):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        loader = tqdm(dataloader, desc="Training", leave=False)
        for images, labels in loader:
            # Map 'normal' (2) to 0, 'glaucoma' (0,1) to 1
            labels = (labels != 2).float().unsqueeze(1).to(device)
            images = images.to(device)
            outputs = model(images)
            loss = criterion(outputs, labels)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * images.size(0)
            predicted = (outputs >= 0.5).float()
            correct += (predicted == labels).sum().item()
            total += labels.size(0)
            current_loss = running_loss / total
            current_accuracy = correct / total
            loader.set_postfix(loss=current_loss, accuracy=current_accuracy)
        epoch_loss = running_loss / total
        accuracy = correct / total
        return epoch_loss, accuracy

    def binary_validate(model, dataloader, criterion):
        model.eval()
        running_loss = 0.0
        correct = 0
        total = 0
        loader = tqdm(dataloader, desc="Validation", leave=False)
        with torch.no_grad():
            for images, labels in loader:
                labels = (labels != 2).float().unsqueeze(1).to(device)
                images = images.to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)
                running_loss += loss.item() * images.size(0)
                predicted = (outputs >= 0.5).float()
                correct += (predicted == labels).sum().item()
                total += labels.size(0)
                current_loss = running_loss / total
                current_accuracy = correct / total
                loader.set_postfix(loss=current_loss, accuracy=current_accuracy)
        epoch_loss = running_loss / total
        accuracy = correct / total
        return epoch_loss, accuracy

    train_loss, train_acc = binary_train(model, train_loader, criterion, optimizer)
    val_loss, val_acc = binary_validate(model, val_loader, criterion)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    train_accuracies.append(train_acc)
    val_accuracies.append(val_acc)

    print(f"Epoch [{epoch+1}/{epochs}] - "
          f"Train Loss: {train_loss:.4f}, Train Accuracy: {train_acc:.4f}, "
          f"Val Loss: {val_loss:.4f}, Val Accuracy: {val_acc:.4f}")
```

### Confusion Matrix

```python
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score, f1_score
import seaborn as sns
import matplotlib.pyplot as plt
from tqdm import tqdm
import torch

def test_with_metrics(model, dataloader, class_names):
    model.eval()
    y_true, y_pred = [], []

    loader = tqdm(dataloader, desc="Testing", leave=False)
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)

            outputs = model(images)  # shape [batch_size, num_classes]
            preds = torch.argmax(outputs, dim=1)

            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())

    # Confusion Matrix
    cm = confusion_matrix(y_true, y_pred)

    accuracy  = accuracy_score(y_true, y_pred)
    precision = precision_score(y_true, y_pred, average='weighted')
    recall    = recall_score(y_true, y_pred, average='weighted')
    f1        = f1_score(y_true, y_pred, average='weighted')

    # Plot
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu', xticklabels=class_names, yticklabels=class_names)
    plt.xlabel("Predicted")
    plt.ylabel("Actual")
    plt.title("Confusion Matrix")

    plt.show()

    print(f"Test Accuracy : {accuracy:.4f}")
    print(f"Precision     : {precision:.4f}")
    print(f"Recall        : {recall:.4f}")
    print(f"F1 Score      : {f1:.4f}")
```

```python
class_names = ['advanced', 'early', 'normal']
test_with_metrics(model, test_loader, class_names)
```
